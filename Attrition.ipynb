```python
# correlation test continue vs continues
```

```python
from scipy import stats
```

```python
def corr(d1,d2):
  corr,p_value=stats.pearsonr(d1,d2)
  if p_value<0.03:
    return False
  else:
    return True
```

```python
corr(df["Age"],df["TotalWorkingYears"])
```

```python
# 2 sample test---> continuous vs continuous
```

```python
def twosample(d1,d2):
  f=0
  t=0
  sample1=d1.sample(frac=0.05)
  sample2=d2.sample(frac=0.05)
  t_test,p_value=stats.ttest_ind(sample1,sample2)
  if p_value < 0.02:
      f=f+1
  else:
      t=t+1
  if f>t:
    return False
  else:
    return True
```

```python
twosample(df["Age"],df["PercentSalaryHike"])
```

```python
#chisquare test catagory vs catagory
```

```python
def chisquare(d1,d2):
  crosstab=pd.crosstab(d1,d2)
  chi2,p_value,dof,E_values=stats.chi2_contingency(crosstab)
  if p_value<0.03:
    return True
  else:
    return False
```

```python
chisquare(df["Attrition"],df["Department"])
```

```python
# continous vs catagory -->F test or Annova test
```

```python
def annova(d1,d2):
  group=df[d2].unique()
  data={}
  for i in group:
    data[i]=df[df[d2]==i][d1].values
  f_value,p_value=stats.f_oneway(*[i for i in data.values()])
  if p_value<0.03:
    return False
  else:
    return True
```

```python
annova("MonthlyIncome","Department")
```
